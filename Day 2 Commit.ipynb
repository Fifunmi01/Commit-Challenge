```python
import numpy as np
import pandas as pd
#NumPy: Generally used for numerical computing and handling multi-dimensional arrays.

#Pandas: Used for data manipulation, analysis, and cleaning (working with structured data like spreadsheets).

#Step 3: Load the dataset
df = pd.read_csv("data.csv")

#Head- Display the first 5 rows to check if it loaded correctly
df.head()


#Step 4
# 1. Shows the size of the dataset (Rows, Columns)
df.shape

# 2. Lists all the column names
df.columns

# 3. Shows data types and if there are any missing values
df.info()

# 4. Shows summary statistics (like average, min, max) for numerical columns
df.describe()

#Observations
#1. Dataset Structure (Rows & Columns)
#The dataset contains 6 columns: index, Duration, Date, Pulse, Maxpulse, and Calories.
#Looking at the shape output, the dataset has 5 rows."


#2. Data Types
#The data consists of a mix of numerical and text data.
#Specifically, Duration, Pulse, and Maxpulse appear to be integers (whole numbers), while Calories is a float (decimal number).
#The Date column is stored as an object (text/string)."

#3. Data Content & Context
#The data appears to represent fitness or workout sessions.
#The Duration column likely tracks how long the exercise lasted (e.g., 60 or 45 minutes), while Pulse and Maxpulse track heart rate intensity.
#The Date column shows these sessions took place in December 2020."

#Step 5
# 1.Numpy Array using 'Pulse' column and converting it to a NumPy array
pulse_array = np.array(df['Pulse'])

# 2. Perform basic NumPy operations
print("Mean Pulse:", np.mean(pulse_array))  # The average pulse
print("Max Pulse:", np.max(pulse_array))    # The highest pulse recorded
print("Min Pulse:", np.min(pulse_array))    # The lowest pulse recorded

#Step 6
# Check for missing values in every column
# isnull() finds the empty cells, and sum() counts them
print(df.isnull().sum())

#This is the result of df.isnull().sum(). It tells you:
#Date: Has 1 missing value.
#Calories: Has 2 missing values.
#Duration, Pulse, Maxpulse: Have 0 missing values (they are complete).

#Answers for Step 6
#1. Are there missing values? Yes.
#2. Which column has the most? Calories (it has 2, while Date only has 1).

#Step 7
# Filter data to show rows where Calories are greater than 300
filtered_df = df[df['Calories'] > 300]

# Display the first few rows of the filtered data
filtered_df.head()

#Step 8
#1. What was easy?
#Example: "Loading the dataset and running basic commands like df.head() was straightforward."

#2.What was confusing?
#.Example: "Understanding how to select columns for filtering took a bit of trial and error." (Or whatever actually confused you!)

#3.One thing you learned today.
#Example: "I learned how to use isnull().sum() to quickly find missing data in a dataset."



```
