```python
#Loading the Necessary Packages Necessary
import pandas as pd
import matplotlib.pyplot as plt # This will be needed for Drawing the Plots
import seaborn as sns #This is needed for the Chart and making the Chart more visible

# Loading  and cleaning the data (This is the Previous code from the Day 2)
df = pd.read_csv("data.csv")

df = df.drop_duplicates()

df = df.dropna(subset=['Date'])

df['Calories'] = df['Calories'].fillna(df['Calories'].mean())

df['Date'] = pd.to_datetime(df['Date'], errors='coerce')

df = df.dropna(subset=['Date'])

```

```python
# DAY 3 CODE
# Sorts the data from oldest date to newest date
df = df.sort_values(by='Date')

# Setting a nice visual theme for our charts I think the white background with grey lines will do the trick
sns.set_theme(style="whitegrid")

# This is just to print that the creation of visuals will go well and Also to show An Output at least
print("Data is clean and ready for visualization!")
```

```python
# Step 2: Creation of the Line Chart
#Creates a blank canvas for our chart. 'figsize=(12, 5)' makes it 12 inches wide and 5 inches tall.
plt.figure(figsize=(12, 5))

# Plot Calories over Date
sns.lineplot(x='Date', y='Calories', data=df, marker='o', color='crimson')

# This is the Full Explanation of the Code
# x='Date': Puts Dates on the bottom axis.
# y='Calories': Puts Calories on the left vertical axis.
# data=df: Tells it to pull this information from our cleaned dataset 'df'.
# marker='o': Puts a visible dot ('o') at every individual data point on the line.
# color='crimson': Colors the line dark red.

#Putting in the Parameters like The Titles and the Label bascially the Line Graph
plt.title('Trend of Calories Burned Over Time', fontsize=14) # Adds a large title to the very top of the chart.
plt.xlabel('Date', fontsize=12) # Adds a label to the bottom axis so people know what the numbers mean.
plt.ylabel('Calories Burned', fontsize=12) # Adds a label to the side axis.
plt.xticks(rotation=45) # Tilts the dates so they are easier to read
plt.tight_layout() # This adjusts the margins of the Picture


# The final command that actually pushes the finished picture to your screen.
plt.show()
```

```python
# Step 3: Creation of the Bar Chart
# Creates a blank canvas for our chart. 'figsize=(8, 5)' makes it 8 inches wide and 5 inches tall.
plt.figure(figsize=(8, 5))

# Plotting Calories over Duration
sns.barplot(x='Duration', y='Calories', data=df, errorbar=None, hue='Duration', palette='viridis', legend=False)

# This above code tells Seaborn to draw a bar chart.
# x='Duration': Puts the workout lengths (30, 45, 60 mins) on the bottom.
# y='Calories': Puts the Calories on the side.
# data=df: Uses our dataset.
# errorbar=None: Removes the little black "error" lines that Seaborn adds by default, keeping the chart looking clean.
# palette='viridis': Uses a pre-made set of colors called 'viridis' (a mix of greens, blues, and yellows) for the bars.


# Adding  the Title, Bottom Label, and Side Label.
plt.title('Average Calories Burned per Workout Duration', fontsize=14)
plt.xlabel('Workout Duration (Minutes)', fontsize=12)
plt.ylabel('Average Calories', fontsize=12)

# Displays the finished bar chart.
plt.show()
```

```python
# Step 4: Creation of the Histogram
# Creates a blank canvas for our chart. 'figsize=(8, 5)' makes it 8 inches wide and 5 inches tall.
plt.figure(figsize=(8, 5))

# Plot the distribution of Pulse rates
sns.histplot(df['Pulse'], bins=10, kde=True, color='purple')

# This code above tells  Seaborn to draw a histogram.
# df['Pulse']: Notice we didn't use x= or y= here. Histograms only need ONE column of data to work! We give it the Pulse column.
# bins=10: This divides our data into 10 groups (e.g., pulses from 90-100, 100-110, etc.) so we can count them.
# kde=True: Stands for Kernel Density Estimate. It draws a smooth, curved line over the bars to show the overall shape of the data.
# color='purple': Makes the bars purple.

# Adding the Title, Bottom Label, and Side Label.
plt.title('Distribution of Average Pulse Rates', fontsize=14)
plt.xlabel('Pulse Rate', fontsize=12)
plt.ylabel('Frequency (Number of Workouts)', fontsize=12)

# Displaying the finished histogram
plt.show()
```

Step 5: Chart Explanations & Insights:

1. Line Chart (Trend Analysis): I used a line chart to visualize the trend of Calories burned over time (Date).

- Insight/Trend : The trend shows daily fluctuations in calorie expenditure, with noticeable spikes on specific days (like 2020/12/02), indicating days with much higher-intensity workouts.

2. Bar Chart (Category Comparison): I created a bar chart to compare the average Calories burned across different workout Duration categories.

- Insight/Trend : There is a clear relationship: The longer the workout duration (e.g., 60 minutes vs. 30 minutes), the higher the average calories burned.

3. Histogram (Data Distribution): I used a histogram with a KDE curve to view the distribution of the Pulse column.

- Insight/Trend : The distribution shows that the majority of the heart rate readings are concentrated roughly between 100 and 110 beats per minute, which is the most frequent pulse zone for these workouts.
