```python
#Loading The required Packages necessary
import pandas as pd
import numpy as np

# Step 1: Loading the dataset
df = pd.read_csv("data.csv")

# Displaying the first few rows
df.head()
```

```python
# Step 2: Checking Duplicates: In order to do this you must count how many duplicates exist
duplicate_count = df.duplicated().sum()
print(f"Number of duplicate rows: {duplicate_count}")

# Removing  the duplicates
df.drop_duplicates(inplace=True)

#Printing the Successful outputs
print("Duplicates removed successfully!")
```

```python
# Step 3: There are missing values in the Date Column so we will have to drop rows where 'Date' is missing
df.dropna(subset=['Date'])

# Calculate the mean of the 'Calories' column
mean_calories = df['Calories'].mean()

# Fill missing 'Calories' with the mean value
df['Calories'].fillna(mean_calories)

# Verifing that there are no more missing values
print(df.isnull().sum())
```

```python
# Step 4: Convert the 'Date' column to datetime format
# errors='coerce' turns completely unreadable dates into 'NaT' (Not a Time)
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')

# Drop any rows that became 'NaT' because the date format was too broken to read
df = df.dropna(subset=['Date'])

# Show the updated data types to confirm the change
print(df.dtypes)
```

Step 5: Explanation of the Decisions

Data Cleaning Decisions:
Duplicates: I used df.duplicated().sum() to identify duplicates and df.drop_duplicates() to remove them, ensuring they don't disturb  our analysis.

Missing Values: I dropped rows with missing dates because having a workout without a date  format is incomplete.

For missing calories, I filled the empty cells with the mean (fillna()) to retain the rest of the row's valuable data.

Data Types: I converted the Date column from an object (string) to a datetime format using pd.to_datetime() so it can be used for proper time-series analysis in the future.
